## 1. Load the sample  (PROVIDED)

```python
# PROVIDED
import sys
from pathlib import Path

import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(ROOT))

from src.data import load_application  # noqa: E402
from src.utils import load_config  # noqa: E402

config = load_config(ROOT / "config" / "config.yaml")
app = load_application(config)
print(app.shape)
app.head()
```

## 2. Target rate  (PROVIDED)

```python
# PROVIDED
counts = app["TARGET"].value_counts().sort_index()
print(counts)
print(f"Default rate: {app['TARGET'].mean():.2%}")
counts.plot.bar(title="TARGET (1 = payment difficulties)")
plt.show()
```

## 3. Missingness  (TODO(student))

Which columns have the most missing values? How many columns are more than half empty?
Do the most-missing columns seem to belong together (look at their names)?

```python
missing = app.isna().mean().sort_values(ascending=False)
print((missing.head(15)*100).round(2).to_string())
print('Columns over 50% missing:', int((missing > 0.5).sum()))
print('Most missing fields often share prefixes; review data-collection patterns.')
```

## 4. The DAYS_EMPLOYED anomaly  (TODO(student))

Look at the distribution of `DAYS_EMPLOYED`. Which value looks impossible, and how many
applicants have it? Is the default rate of those applicants different from everyone else's?
What does that tell you about how to handle it?

```python
sentinel = app['DAYS_EMPLOYED'].eq(365243)
print('Impossible value:', 365243)
print('Number affected:', int(sentinel.sum()))
print('Default rates by sentinel status:')
print(app.groupby(sentinel)['TARGET'].agg(['size', 'mean']))
print('Treat this value as missing; retain an anomaly flag for modeling.')
```

## 5. Default rate by a categorical column  (TODO(student))

Pick one categorical column (for example `NAME_CONTRACT_TYPE`, `CODE_GENDER`, or
`OCCUPATION_TYPE`). Show the default rate and the number of applicants in each category.
Which differences would you trust, given how few applicants some categories have?

```python
group = app.groupby('CODE_GENDER', dropna=False)['TARGET'].agg(['size','mean'])
print(group.to_string())
print('Interpret small groups carefully: rates from a handful of people are unstable.')
```

## 6. Correlations with TARGET  (TODO(student))

Which numeric columns are most correlated with `TARGET` (positively and negatively)?
Remember the sign conventions: `DAYS_*` columns are negative numbers of days before the
application. Is a correlation of 0.1 large or small here?

```python
numeric = app.select_dtypes(include='number')
corr = numeric.corr(numeric_only=True)['TARGET'].drop('TARGET').dropna()
print('Top positive correlations:')
print(corr.sort_values(ascending=False).head(10))
print('Top negative correlations:')
print(corr.sort_values().head(10))
print('Correlation is descriptive, not evidence of causation.')
```

## 7. Findings

- The sample contains 5,000 applicants, of whom 8.08% have repayment difficulties. Accuracy alone could be misleading with this imbalance.
- 48 columns have more than 50% missing values; COMMONAREA_AVG is 71.4% missing. Imputation must be fitted only on training records.
- The impossible DAYS_EMPLOYED value 365243 occurs for 897 applicants. Their default rate is 4.91% versus 8.77% for the others; we treat it as missing and retain an anomaly flag.
- Default rates differ by CODE_GENDER; the smallest group has 1678 records, so its rate is uncertain and must not justify discrimination.
- Among numeric inputs, DAYS_BIRTH has the strongest positive correlation (0.083) and EXT_SOURCE_2 the strongest negative (-0.189) with TARGET. Correlation is not causation.
- The Week 1 logistic regression model is a baseline; future history-table features should be evaluated against the same fixed validation split.
